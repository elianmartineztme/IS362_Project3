```python

import sqlite3
import pandas as pd

# Connect to the Chinook database
from pathlib import Path

db_path = Path("Chinook_Sqlite (1).sqlite")

if not db_path.is_file():
    raise FileNotFoundError("Chinook database file not found.")

conn = sqlite3.connect(str(db_path))

# Check the available tables
tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

display(tables)
```

```python

# Combine customer purchases with tracks and albums

query = """
SELECT
    c.LastName,
    c.FirstName,
    t.Name,
    a.Title
FROM Customer AS c
JOIN Invoice AS i
    ON c.CustomerId = i.CustomerId
JOIN InvoiceLine AS il
    ON i.InvoiceId = il.InvoiceId
JOIN Track AS t
    ON il.TrackId = t.TrackId
JOIN Album AS a
    ON t.AlbumId = a.AlbumId
ORDER BY
    c.LastName,
    c.FirstName;
"""

# Display results in a pandas DataFrame
df = pd.read_sql_query(query, conn)

df.head()
```

```python

# Count unique customers in the purchase results
unique_customers = df[["LastName", "FirstName"]].drop_duplicates()

print("Total purchased tracks:", len(df))
print("Unique customers:", len(unique_customers))
```
